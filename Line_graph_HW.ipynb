```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
July = {'Day': ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday'], 'Births': [12, 15, 11, 9, 1, 9, 21]}
df = pd.DataFrame(July, columns = ['Day', 'Births'])
df.head(7)
```

```python
sns.lineplot(x=df['Day'], y=df['Births'], color='red', marker='o')
plt.title('July Birthdays')
plt.xlabel('Day')
plt.ylabel('Births')
plt.grid(True)
plt.show()
```

```python
August = {'Day': ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday'], 'Births': [17, 5, 2, 11, 1, 8, 29]}
df_A = pd.DataFrame(August, columns = ['Day', 'Births'])
df_A.head(7)
```

```python
sns.lineplot(x=df_A['Day'], y=df_A['Births'], color='green', marker='x')
plt.title('August Birthdays')
plt.xlabel('Day')
plt.ylabel('Births')
plt.grid(True)
plt.show()
```

```python
df_july = df.copy()
df_july['Month'] = 'July'

df_august = df_A.copy()
df_august['Month'] = 'August'

df_combined = pd.concat([df_july, df_august], ignore_index=True)

plt.figure(figsize=(10, 6))
sns.lineplot(x='Day', y='Births', hue='Month', data=df_combined, marker='o')
plt.title('July and August Birthdays')
plt.xlabel('Day')
plt.ylabel('Births')
plt.grid(True)
plt.show()
```

```python
x_values = np.arange(1, 11) # x from 1 to 10
y_values = 6 * (x_values**2) + x_values + 1

eq_df = pd.DataFrame({'x': x_values, 'y': y_values})

plt.figure(figsize=(10, 6))
sns.lineplot(x='x', y='y', data=eq_df, marker='o', color='blue')
plt.title('Solution of y = 6x^2 + x + 1')
plt.xlabel('x')
plt.ylabel('y')
plt.xticks(x_values) # Ensure all x values are shown on the axis
plt.grid(True)
plt.show()
```
